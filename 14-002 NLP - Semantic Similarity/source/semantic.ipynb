## 1. Compare individual words (lesson extract)

```python
import spacy

# Load the medium English language model.
nlp = spacy.load("en_core_web_md")

# Process the three words.
word1 = nlp("cat")
word2 = nlp("monkey")
word3 = nlp("banana")

# Compare each pair of words.
print("cat / monkey:", word1.similarity(word2))
print("banana / monkey:", word3.similarity(word2))
print("banana / cat:", word3.similarity(word1))
```

### Observation:
The **cat–monkey** comparison is relatively strong because both words refer to animals. **Monkey–banana** also shows a meaningful association, which may reflect how the words are used together in the model’s training data. **Cat–banana** is generally less similar because the words refer to less closely related concepts. These are statistical similarity scores, not proof that the model understands the relationship. The precise scores depend on the installed model version.

## 2. Compare all pairs of words (lesson extract)

```python
tokens = nlp("cat apple monkey banana ")

for token1 in tokens:
    for token2 in tokens:
        print(token1.text, token2.text, token1.similarity(token2))
```

### Observation:
The word **apple** is generally more similar to **banana** than to **cat**, because both apple and banana are fruits. A word compared with itself produces a similarity of approximately **1.0**. The comparisons also show that similarity is not necessarily transitive: a high score for one pair does not imply a high score for another pair.

## 3. Additional Example

```python
examples = ["doctor", "nurse", "mountain"]
for first in examples:
    for second in examples:
        if first != second:
            print(first, second, round(nlp(first).similarity(nlp(second)), 3))
```

### Observation:
The words *doctor* and *nurse* are expected to have a strong semantic relationship because they refer to professions in the healthcare industry. However, the results obtained show a similarity score of 0.114 for doctor and nurse, compared with 0.154 for doctor and mountain. This demonstrates that the model's similarity scores do not always correspond with the relationships humans would expect between words.

## 4. Compare sentences (lesson extract)

```python
sentence_to_compare = "Why is my cat on the car"

sentences = [
    "where did my dog go",
    "Hello, there is my car",
    "I've lost my car in my car",
    "I'd like my boat back",
    "I will name my dog Diana",
]

model_sentence = nlp(sentence_to_compare)

for sentence in sentences:
    similarity = nlp(sentence).similarity(model_sentence)
    print(sentence + " - " + str(similarity))
```

### Observation
Sentences that share relevant words or meanings with the target sentence may receive higher similarity scores. However, a score is only an estimate of semantic relatedness; it does not establish that two sentences have the same meaning.
